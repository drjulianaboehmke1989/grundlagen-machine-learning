# Aufgabe 3: Lineare Regression mit dem `wave`-Datensatz

*Grundlagen des Maschinellen Lernens · Übungsaufgaben Teil 3*

In dieser Übung untersuchen Sie den synthetischen **wave-Datensatz** aus Müller und Guido (*Introduction to Machine Learning with Python*). Der Datensatz enthält **ein numerisches Merkmal** und eine **metrische Zielvariable**. Sie entwickeln selbstständig einen vollständigen Arbeitsablauf zur linearen Regression.

## Aufgabe 3.1: Datensatz bereitstellen und untersuchen

Der Datensatz wird über die Bibliothek **mglearn** bereitgestellt. Verwenden Sie den `wave`-Datensatz mit **40 Beobachtungen**.

a) Laden bzw. erzeugen Sie den Datensatz und bestimmen Sie die Anzahl der Beobachtungen und Merkmale.

b) Untersuchen Sie, in welcher Form die Eingabedaten und die Zielvariable vorliegen. Welche Form benötigt das Regressionsmodell?

c) Lassen Sie sich einige Beobachtungen anzeigen und beschreiben Sie den Wertebereich der Eingabe- und Zielvariablen.

## Aufgabe 3.2: Explorative Datenanalyse

b) Beschreiben Sie den Zusammenhang zwischen dem Merkmal und der Zielvariable. Ist ein linearer Zusammenhang plausibel? Begründen Sie Ihre Einschätzung anhand der Abbildung.

c) Prüfen Sie die Daten auf auffällige Beobachtungen und erläutern Sie, welche Bedeutung diese für eine lineare Regression haben könnten.
