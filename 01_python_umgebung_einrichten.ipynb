# Python-Umgebung einrichten

Dieses Notebook zeigt, wie eine **virtuelle Python-Umgebung** angelegt und als Jupyter-Kernel registriert wird. Die eigentliche Umgebung wird im Terminal erstellt; anschließend kann dieses Notebook zur Kontrolle der Installation genutzt werden.
Eine Umgebung hilft dabei die benötigten Versionen von Paketen zu reproduzieren. Unterschiedliche Versionen von Paketen haben unterschiedliche Anforderungen, Syntax und Verknüpfungen untereinander. 

## 1. Warum eine virtuelle Umgebung?

Eine virtuelle Umgebung erlaubt es Python-Pakete projektbezogen aufzusetzen. Dadurch können verschiedene Projekte unterschiedliche Paketversionen verwenden, ohne sich gegenseitig zu beeinflussen. Für reproduzierbare Machine-Learning-Projekte ist dies ein grundlegender Bestandteil der Arbeitsumgebung.

## 2. Umgebung im Terminal erstellen

### Windows (PowerShell)
```powershell
py -m venv .venv
.\.venv\Scripts\Activate.ps1
python -m pip install --upgrade pip
pip install numpy pandas matplotlib scikit-learn jupyter ipykernel
python -m ipykernel install --user --name ml-kurs --display-name "Python (ml-kurs)"
```

### macOS / Linux
```bash
python3 -m venv .venv
source .venv/bin/activate
python -m pip install --upgrade pip
pip install numpy pandas matplotlib scikit-learn jupyter ipykernel
python -m ipykernel install --user --name ml-kurs --display-name "Python (ml-kurs)"
```

Danach in Jupyter unter **Kernel** den Eintrag **Python (ml-kurs)** auswählen. Jupyter muss erst neu gestartet werden.

## 3. Aktive Python-Umgebung prüfen

````python
import sys
from pathlib import Path

print("Python-Version:", sys.version.split()[0])
print("Python-Pfad:", sys.executable)
print("Arbeitsverzeichnis:", Path.cwd())
````

## 4. Benötigte Bibliotheken prüfen

````python
import numpy as np
import pandas as pd
import matplotlib
import sklearn

print("NumPy:", np.__version__)
print("pandas:", pd.__version__)
print("Matplotlib:", matplotlib.__version__)
print("scikit-learn:", sklearn.__version__)
````

## 6. Reproduzierbarkeit: Paketliste speichern

Im Terminal kann der aktuelle Paketstand dokumentiert werden:

```bash
pip freeze > requirements.txt
```

Eine Umgebung kann später auf Basis dieser Datei rekonstruiert werden:

```bash
pip install -r requirements.txt
```

## 7. Mini-Check Lineare Regression

````python
from sklearn.linear_model import LinearRegression

X = [[1], [2], [3], [4]]
y = [3, 5, 7, 9]
model = LinearRegression().fit(X, y)
print("Achsenabschnitt:", model.intercept_)
print("Steigung:", model.coef_[0])
````

**Erwartung:** Der Test sollte ungefähr einen Achsenabschnitt von 1 und eine Steigung von 2 ergeben. Wenn die Zelle fehlerfrei ausgeführt wird, ist die grundlegende Kursumgebung einsatzbereit.
